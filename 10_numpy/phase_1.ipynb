## Numpy Array and basics


```python
import numpy as np 
import time
```

### create array from list


```python
arr_1=np.array([1,2,3,4,5])
print("1D Array : ", arr_1)


arr_2=np.array([[1,2,3],[4,5,6]])
print("2D Array : ",arr_2)

```

### List vs numpy array


```python
py_list=[1,2,3,4]
print("Python list mutiplication ", py_list*2)

np_array=np.array([1,2,3])
print("Python array  mutiplication ", np_array*2)

start=time.time()
py_list=[i*2 for i in range(1000000)]
print("\n List opertion time: ",time.time()-start)

start=time.time()
np_array=np.arange(100000)*2
print("\n Numpy opertion time: ",time.time()-start)
```

### creating array from scratch


```python
zeros=np.zeros((3,4))
print("zeros array :\n",zeros)

ones=np.ones((2,3))
print("ones array :\n",ones)

full= np.full((2,2),7)
print("constant  array :\n",full)

random=np.random.random((2,3))
print("random array :\n",random)

sequence=np.arange(0,10,2)
print("sequence array :\n",sequence)
```

### Vector,Matrix and Tensor

Scalar → 0D
   
Vector → 1D
   
Matrix → 2D
   
Tensor → 3D or higher

```python
vector=np.array([1,2,3])
print("Vector : ",vector)

matrix=np.array([[1,2,3],[4,5,6]])
print("Matrix : ",matrix)

tensor=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print("Tensor : ",tensor)

```

### Array properties

```python
arr=np.array([[1,2,3],[4,5,6]])
print(arr)
print("Shape " ,arr.shape)
print("Dimension ",arr.ndim)
print("Size ",arr.size)
print("Dtype ",arr.dtype)
```

### Array Reshaping

```python
arr=np.arange(12)
print("Original Array ",arr)

reshaped=arr.reshape((4,3))
print("\nReshaped Array ", reshaped)


flattened=reshaped.flatten()
print("\nFlatten Array ",flattened)

# ravel (return view ,instead of copy)
raveled=reshaped.ravel()
print("\nRaveled ",raveled)
raveled[0] = 100
#flattened[0]=100

print(reshaped)

## Tranpose
transpose =reshaped.T
print("\n Tranpose",transpose)

```
