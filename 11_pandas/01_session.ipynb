### Handling Null Values

```python
coffee.loc[[2,3],['Units Sold']]=np.nan
coffee
```

```python
# to check NA count
coffee.isna().sum()
```

```python
# fill "NA" value
coffee=coffee.fillna(coffee['Units Sold'].mean())
coffee
```

```python
coffee.loc[[0,1],['Units Sold']]=15
coffee
```

```python
# interpolate() is a Pandas method used to fill missing (NaN) values by estimating values based on the surrounding known values.
coffee["Units Sold"]=coffee["Units Sold"].interpolate()
coffee
```

```python
# drop 
# coffee.dropna()
# or 
coffee=coffee.dropna(subset=["Units Sold"])
```

```python
coffee[coffee['Units Sold'].isna()]
coffee[coffee['Units Sold'].notna()]
```

## Advanced Functionality 
### .shift() .rank() .rolling() .cumsum()

```python
# shift(1)-> previous row value
# shift(-1)-> next row value
coffee['yesterday_revenue']=coffee['revenue'].shift(2)
```

```python
# rank()
bios['height_rank']=bios['height_cm'].rank(ascending=True)
```

```python
bios.sort_values(['height_rank']).sample(10)[['name','height_rank']]

# head()-> first n rows 
# sample()->  show n random rows
```

```python
# cumsum
coffee['cumulative_revenue']=coffee['revenue'].cumsum()
```

```python
#rolling 

latte=coffee[coffee['Coffee Type']=='Latte'].copy()
latte['3day']=latte['Units Sold'].rolling(3).sum()
print(latte)
```

```python
import pandas as pd

# Creating a DataFrame
data = {
    'Date': ['2024-05-01', '2024-05-01', '2024-05-01', '2024-05-02', '2024-05-02', '2024-05-03', '2024-05-03', '2024-05-03'],
    'Item': ['Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Orange', 'Apple', 'Orange'],
    'Units Sold': [30, 21, 15, 40, 34, 20, 45, 25],
    'Price Per Unit': [1.0, 0.5, 0.75, 1.0, 0.5, 0.75, 1.0, 0.75],
    'Salesperson': ['John', 'John', 'John', 'Alice', 'Alice', 'John', 'Alice', 'John']
}

df = pd.DataFrame(data)

# Display the DataFrame
df
```

```python

pivot_table = pd.pivot_table(df, values='Units Sold', index='Date', columns='Item', aggfunc='sum')
pivot_table
```

```python
# pivot()
# - Reshapes data from long format to wide format.
# - Does NOT perform aggregation.
# - Each index + column combination must be unique.
# - Duplicate combinations → ValueError.

df.pivot(
    index='Day',
    columns='Coffee Type',
    values='Units Sold'
)


# pivot_table()
# - Reshapes data AND performs aggregation.
# - Handles duplicate index + column combinations.
# - Uses aggfunc to decide how to combine values.
# - Default aggfunc = 'mean'.

df.pivot_table(
    index='Day',
    columns='Coffee Type',
    values='Units Sold',
    aggfunc='sum'
)

# Common aggfunc:
# 'sum', 'mean', 'count', 'min', 'max'
```
